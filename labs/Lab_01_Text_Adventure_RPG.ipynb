## Part 2: Building the Character System 

Let's start with the foundation: characters! Every RPG needs a way to represent the player and enemies.

### 2.1 The Character Base Class

Create a `Character` class with:
- **Attributes**: `name`, `health`, `max_health`, `strength`, `defense`
- **Methods**:
  - `__init__()`: Initialize the character
  - `is_alive()`: Returns True if health > 0
  - `take_damage(amount)`: Reduce health (but don't go below 0)
  - `attack(target)`: Roll dice, calculate damage, apply to target
  - `__str__()`: Return a nice string representation

### The D20 Combat System

We'll use a simple dice-based combat system:

1. **Attack Roll**: Roll a d20 (1-20). Add attacker's strength.
2. **Defense Check**: If attack roll > target's defense, hit! Otherwise, miss.
3. **Damage**: On a hit, deal `strength` damage (or roll dice for variable damage)

```python
import random

def roll_d20():
    return random.randint(1, 20)

def roll_dice(num_dice, sides):
    """Roll multiple dice and return the total. E.g., roll_dice(2, 6) for 2d6."""
    return sum(random.randint(1, sides) for _ in range(num_dice))
```

````python
import random

def roll_d20():
    """Roll a 20-sided die."""
    return random.randint(1, 20)

def roll_dice(num_dice, sides):
    """Roll multiple dice and return the total. E.g., roll_dice(2, 6) for 2d6."""
    return sum(random.randint(1, sides) for _ in range(num_dice))


class Character:
    """Base class for all characters in the game."""

    def __init__(self, name, health, strength, defense):
        self.name       = name
        self.health     = health
        self.max_health = health
        self.strength   = strength
        self.defense    = defense

    def is_alive(self):
        """Return True if the character still has health."""
        return self.health > 0

    def take_damage(self, amount):
        """Reduce health by amount, clamped to 0."""
        self.health = max(0, self.health - amount)
        print(f"  {self.name} takes {amount} damage! ({self.health}/{self.max_health} HP remaining)")

    def attack(self, target):
        """D20 attack roll vs target's defense; deal 1d6+STR on hit."""
        roll   = roll_d20()
        total  = roll + self.strength
        print(f"  {self.name} attacks {target.name}! (rolled {roll} + {self.strength} STR = {total} vs DEF {target.defense})")
        if total >= target.defense:
            damage = roll_dice(1, 6) + self.strength
            print(f"  Hit! Deals {damage} damage.")
            target.take_damage(damage)
        else:
            print(f"  Miss!")

    def _hp_bar(self, width=20):
        """Return a simple ASCII HP bar."""
        filled = int((self.health / self.max_health) * width)
        return f"[{'#' * filled}{'.' * (width - filled)}]"

    def __str__(self):
        return f"{self.name} {self._hp_bar()} {self.health}/{self.max_health} HP"
````

### 2.2 Player and Enemy Subclasses

Now create subclasses that inherit from `Character`:

**Player class** might have:
- An `inventory` list
- Methods: `pick_up(item)`, `use_item(item_name)`, `show_inventory()`
- A special ability (heal, power attack, etc.)

**Enemy class** might have:
- An `xp_value` (experience points when defeated)
- A `loot` attribute (what they drop)
- Override `attack()` with different behavior

**Specific enemy types** (Minion, Elite, Boss) can have:
- Different base stats
- Unique abilities
- Different attack messages (make it fun!)

````python
class Player(Character):
    """The player character — carries items and earns XP."""

    STR_UPGRADE_COST  = 20   # XP cost per +1 STR
    HP_UPGRADE_COST   = 10   # XP cost per +1 max HP
    HEAL_COST         = 10   # XP cost to recover 10 HP
    MAX_BONUS         = 8    # cap on bonus STR and bonus max HP

    def __init__(self, name):
        super().__init__(name, health=30, strength=5, defense=12)
        self.inventory   = []
        self.xp          = 0
        self.str_bonus   = 0   # bonus STR earned via XP (max MAX_BONUS)
        self.hp_bonus    = 0   # bonus max HP earned via XP (max MAX_BONUS)

    def pick_up(self, item):
        self.inventory.append(item)
        print(f"  Picked up: {item.name}")

    def use_item(self, item_name):
        item_name_lower = item_name.lower()
        for item in self.inventory:
            if item_name_lower in item.name.lower():
                item.use(self)
                self.inventory.remove(item)
                return
        print(f"  You don't have '{item_name}'.")

    def show_inventory(self):
        if not self.inventory:
            print("  Inventory is empty.")
        else:
            print("  Inventory:")
            for item in self.inventory:
                print(f"    - {item.name}: {item.description}")

    def gain_xp(self, amount):
        self.xp += amount


# ---------------------------------------------------------------------------
# Item Classes
# ---------------------------------------------------------------------------

class Item:
    """Base class for all items."""
    def __init__(self, name, description):
        self.name        = name
        self.description = description

    def use(self, character):
        print(f"  Used {self.name}.")

class HealthPotion(Item):
    def __init__(self):
        super().__init__("Health Potion", "Restores 15 HP.")
    def use(self, character):
        heal = min(15, character.max_health - character.health)
        character.health += heal
        print(f"  {character.name} drinks the potion and recovers {heal} HP!  ({character.health}/{character.max_health})")

class CoffeeMug(Item):
    def __init__(self):
        super().__init__("Coffee Mug", "Stale campus coffee. Restores 8 HP.")
    def use(self, character):
        heal = min(8, character.max_health - character.health)
        character.health += heal
        print(f"  {character.name} gulps the coffee and recovers {heal} HP!  ({character.health}/{character.max_health})")

class BaseballBat(Item):
    def __init__(self):
        super().__init__("Baseball Bat", "Adds +3 STR permanently.")
    def use(self, character):
        character.strength += 3
        print(f"  {character.name} equips the bat! STR is now {character.strength}.")

class EnergyDrink(Item):
    def __init__(self):
        super().__init__("Energy Drink", "Restores 10 HP. Tastes like battery acid.")
    def use(self, character):
        heal = min(10, character.max_health - character.health)
        character.health += heal
        print(f"  {character.name} slams the energy drink and recovers {heal} HP!  ({character.health}/{character.max_health})")

class FirstAidKit(Item):
    def __init__(self):
        super().__init__("First Aid Kit", "Restores 20 HP.")
    def use(self, character):
        heal = min(20, character.max_health - character.health)
        character.health += heal
        print(f"  {character.name} patches up and recovers {heal} HP!  ({character.health}/{character.max_health})")


# ---------------------------------------------------------------------------
# Enemy Classes
# ---------------------------------------------------------------------------

ZOMBIE_NAMES = ["Chad", "Brad", "Kyle", "Todd", "Brent", "Tanner"]

class Enemy(Character):
    """Base class for all enemies."""
    def __init__(self, name, health, strength, defense, xp_value=10):
        super().__init__(name, health, strength, defense)
        self.xp_value = xp_value
        self.loot     = []

class Zombie(Enemy):
    def __init__(self):
        name = f"Zombie {random.choice(ZOMBIE_NAMES)}"
        super().__init__(name, health=15, strength=3, defense=8, xp_value=10)

    def attack(self, target):
        print(f"  {self.name} groans hungrily...")
        super().attack(target)

class ZombieProfessor(Enemy):
    def __init__(self):
        super().__init__("Zombie Professor", health=25, strength=5, defense=12, xp_value=25)
        self.loot = [HealthPotion()]

    def attack(self, target):
        print(f"  {self.name} shrieks: 'THIS WILL BE ON THE FINAL!'")
        super().attack(target)

class ZombieDean(Enemy):
    def __init__(self):
        super().__init__("Dean of the Dead", health=50, strength=7, defense=15, xp_value=100)
        self.loot = [HealthPotion()]

    def attack(self, target):
        roll  = roll_d20()
        total = roll + self.strength
        print(f"  {self.name} lunges! (rolled {roll} + {self.strength} STR = {total} vs DEF {target.defense})")
        if total >= target.defense:
            if roll >= 18:
                damage = roll_dice(2, 6) + self.strength
                print(f"  CRITICAL STRIKE! Deals {damage} damage!")
            else:
                damage = roll_dice(1, 6) + self.strength
                print(f"  Hit! Deals {damage} damage.")
            target.take_damage(damage)
        else:
            print(f"  Miss — he trips on his robes!")
````

## Part 3: Building the World

Now let's create locations for your adventure!

### 3.1 The Location Class

Each location needs:
- **Attributes**: `name`, `description`, `connections` (dict mapping directions to other locations), `enemies` (list), `items` (list)
- **Methods**:
  - `describe()`: Print the location description and what's here
  - `get_exits()`: Return available directions
  - `add_connection(direction, location)`: Connect to another location
  
**Pro tip**: Store connections as a dictionary: `{"north": cave, "south": village}`

````python
class Location:
    """A location in the game world."""

    def __init__(self, name, description):
        self.name        = name
        self.description = description
        self.connections = {}   # {"north": Location, "south": Location, ...}
        self.enemies     = []
        self.items       = []

    def describe(self):
        """Print room name and flavour text only (Game renders enemies/items/exits)."""
        print(f"\n{'=' * 50}")
        print(f"  {self.name}")
        print(f"{'=' * 50}")
        print(f"  {self.description}")

    def get_exits(self):
        """Return a list of available directions."""
        return list(self.connections.keys())

    def add_connection(self, direction, location):
        """Connect this location to another."""
        self.connections[direction] = location
````

````python
def create_world():
    """
    Build all locations, populate enemies/items, wire up connections.
    Returns (starting_location, admin_building, rooftop).
    admin_building and rooftop are returned so Game can unlock the roof exit
    after the ZombieDean is defeated.

    Map:
        [Dorm] --east--> [Quad] --south--> [Science Building]
                           |
                         north
                           |
                        [Library] --north--> [Admin Building]
                                                    |
                                          up (locked until boss dies)
                                                    |
                                              [Rooftop]
    """
    # --- Create locations ---
    dorm = Location(
        "Dorm Room 204",
        "Your room. Textbooks, dirty laundry, and an unsettling silence outside."
    )
    quad = Location(
        "The Quad",
        "The open courtyard between buildings. Two undead students stumble nearby."
    )
    science = Location(
        "Science Building",
        "Flickering lights. A chemical smell. Something rattles in the lab."
    )
    library = Location(
        "The Library",
        "Shelves knocked over. Pages everywhere. The study carrels are not empty."
    )
    admin = Location(
        "Administration Building",
        "The Dean's office is at the end of the hall. The door is ajar."
    )
    rooftop = Location(
        "Rooftop",
        "Cold air, open sky, and the sound of helicopter blades. You made it."
    )
    rooftop.is_escape = True

    # --- Connect locations (bidirectional where appropriate) ---
    dorm.add_connection("east",  quad)
    quad.add_connection("west",  dorm)

    quad.add_connection("south", science)
    science.add_connection("north", quad)

    quad.add_connection("north", library)
    library.add_connection("south", quad)

    library.add_connection("north", admin)
    admin.add_connection("south", library)
    # 'up' from admin --> rooftop is added by Game after boss is defeated

    # --- Populate enemies ---
    quad.enemies    = [Zombie(), Zombie()]
    science.enemies = [Zombie(), ZombieProfessor()]
    library.enemies = [Zombie(), Zombie()]
    admin.enemies   = [ZombieDean()]

    # --- Populate items (max 5 per room) ---
    # Dorm: safe starting area — grab gear before heading out
    dorm.items    = [CoffeeMug(), BaseballBat(), EnergyDrink()]

    # Quad: first combat zone — small reward visible on arrival
    quad.items    = [HealthPotion(), EnergyDrink()]

    # Science: lab shelves have scattered supplies
    science.items = [EnergyDrink(), FirstAidKit()]

    # Library: well-stocked for long-term survivors
    library.items = [HealthPotion(), EnergyDrink(), FirstAidKit()]

    # Admin: one last item before the boss
    admin.items   = [HealthPotion()]

    return dorm, admin, rooftop


# Quick smoke-test
start, admin_building, rooftop = create_world()
start.describe()
print("\nQuad:")
start.connections["east"].describe()
````

## Part 4: The Combat System

Time to make battles exciting! We'll implement a turn-based combat system using a simple state machine.

### 4.1 Combat States

Combat flows through states. For example, here is one simple way to do it:
1. **PLAYER_TURN**: Player chooses an action
2. **ENEMY_TURN**: Enemy attacks
3. **COMBAT_END**: Someone won (or ran away)

```
┌─────────────────────────────────────────────────────────┐
│                                                         │
│  [PLAYER_TURN] ─── attack ──→ [ENEMY_TURN]              │
│       │                            │                    │
│       │                            │                    │
│       │←─────── (enemy alive) ─────┘                    │
│       │                            │                    │
│       │                            ▼                    │
│       │                      (enemy dead?)              │
│       │                            │                    │
│       └── run away ─────→ [COMBAT_END] ←────────────────┘
│                                                         │
└─────────────────────────────────────────────────────────┘
```

````python
class Combat:
    """Manages turn-based combat between player and enemy."""

    PLAYER_TURN = "player_turn"
    ENEMY_TURN  = "enemy_turn"
    COMBAT_END  = "combat_end"

    def __init__(self, player, enemy):
        self.player = player
        self.enemy  = enemy
        self.state  = Combat.PLAYER_TURN

    def start(self):
        """Run combat until it ends. Returns 'victory', 'defeat', or 'fled'."""
        print(f"\n  {'=' * 50}")
        print(f"  COMBAT: {self.player.name}  vs  {self.enemy.name}")
        print(f"  {'=' * 50}")

        while self.state != Combat.COMBAT_END:
            if self.state == Combat.PLAYER_TURN:
                self.player_turn()
            elif self.state == Combat.ENEMY_TURN:
                self.enemy_turn()

        return self.get_result()

    def player_turn(self):
        """Player chooses: attack / use item / inventory / run."""
        print(f"\n  {self.player}")
        print(f"  {self.enemy}")
        print("\n  Actions: attack  |  use [item]  |  inventory  |  run")
        action = input("  > ").lower().strip()
        parts  = action.split()
        cmd    = parts[0] if parts else ""

        if cmd == "attack":
            self.player.attack(self.enemy)
            if not self.enemy.is_alive():
                print(f"\n  {self.enemy.name} has been defeated!")
                self.state = Combat.COMBAT_END
            else:
                self.state = Combat.ENEMY_TURN

        elif cmd == "use" and len(parts) > 1:
            self.player.use_item(" ".join(parts[1:]))
            self.state = Combat.ENEMY_TURN   # costs your turn

        elif cmd in ["inventory", "i"]:
            self.player.show_inventory()     # free peek

        elif cmd == "run":
            if random.random() < 0.5:
                print("  You manage to escape!")
                self.state = Combat.COMBAT_END
            else:
                print("  Blocked! Cannot escape!")
                self.state = Combat.ENEMY_TURN

        else:
            print("  Unknown action. Try 'attack', 'use [item]', or 'run'.")

    def enemy_turn(self):
        """Enemy takes its turn."""
        print(f"\n  {self.enemy.name}'s turn...")
        self.enemy.attack(self.player)
        if not self.player.is_alive():
            print()
            print("  ╔══════════════════════════════╗")
            print("  ║        Y O U  D I E D !      ║")
            print("  ╚══════════════════════════════╝")
            self.state = Combat.COMBAT_END
        else:
            self.state = Combat.PLAYER_TURN

    def get_result(self):
        """Return 'victory', 'defeat', or 'fled'."""
        if not self.enemy.is_alive():
            return "victory"
        elif not self.player.is_alive():
            return "defeat"
        else:
            return "fled"
````

## Part 5: The Main Game Loop

Now let's tie it all together! The `Game` class manages the overall game state and the main loop.

### 5.1 Game States

Your main game should have these states:
- **EXPLORING**: Player is moving around, examining locations
- **IN_COMBAT**: Player is fighting an enemy
- **GAME_OVER**: Player died
- **VICTORY**: Player won!

### 5.2 Available Actions

When exploring, the player might:
- Move: `go north`, `go south`, `go east`, `go west`
- Look around: `look`
- Fight: `fight` or `attack`
- Inventory: `inventory` or `i`
- Get help: `help`
- Quit: `quit`

Handle invalid input gracefully!

````python
class Game:
    """Main game controller — manages state, world, and player loop."""

    EXPLORING = "exploring"
    GAME_OVER  = "game_over"
    VICTORY    = "victory"

    def __init__(self):
        self.player           = None
        self.current_location = None
        self.admin_building   = None
        self.rooftop          = None
        self.state            = Game.EXPLORING
        self.game_running     = True

    # ------------------------------------------------------------------
    # Setup
    # ------------------------------------------------------------------

    def start(self):
        """Initialize and run the game."""
        self.show_intro()
        self.create_player()
        start_loc, self.admin_building, self.rooftop = create_world()
        self.current_location = start_loc
        self._show_room()

        while self.game_running:
            if self.state == Game.EXPLORING:
                self.exploration_loop()
            elif self.state == Game.GAME_OVER:
                self.show_game_over()
                break
            elif self.state == Game.VICTORY:
                self.show_victory()
                break

    def show_intro(self):
        print("\n" + "=" * 60)
        print("      D E A D   C R E D I T S")
        print("       A Campus Zombie Survival")
        print("=" * 60)
        print("\nThe semester was already brutal — now the faculty have turned.")
        print("Groaning professors stalk the hallways. The Dean himself has")
        print("become something unspeakable.")
        print("\nYour only hope: fight through campus, defeat the Dean of the")
        print("Dead, and reach the rooftop where a rescue helicopter waits.")
        print("=" * 60)

    def create_player(self):
        print("\nWhat's your name, survivor?")
        name = input("  > ").strip() or "Student"
        self.player = Player(name)
        print(f"\nAlright, {name}. Let's try to survive this semester.")
        self.show_help()

    # ------------------------------------------------------------------
    # Room display  (lives in Game so it always reads live location data)
    # ------------------------------------------------------------------

    def _show_room(self):
        """
        Print the current location: name, flavour text, enemies, items, exits.
        Keeping this in Game (not in Location.describe) means it always reads
        the live items/enemies/connections lists directly — no stale instances.
        """
        loc = self.current_location
        print(f"\n{'=' * 50}")
        print(f"  {loc.name}")
        print(f"{'=' * 50}")
        print(f"  {loc.description}")

        # Enemies
        alive = [e for e in loc.enemies if e.is_alive()]
        if alive:
            print(f"\n  Enemies:")
            for e in alive:
                print(f"    - {e.name}  (HP: {e.health}/{e.max_health})")

        # Items on the ground
        if loc.items:
            print(f"\n  Items on the ground:")
            for item in loc.items:
                print(f"    - {item.name}: {item.description}")
        else:
            print(f"\n  No items on the ground.")

        # Exits
        exits = loc.get_exits()
        print(f"\n  Exits: {', '.join(exits) if exits else 'none'}")

    # ------------------------------------------------------------------
    # Exploration
    # ------------------------------------------------------------------

    def exploration_loop(self):
        print("\nWhat do you do?  (type 'help' for commands)")
        command = input("  > ").lower().strip()
        parts   = command.split()
        if not parts:
            return
        action = parts[0]

        if action == "help":
            self.show_help()
        elif action == "look":
            self._show_room()              # always reads live data from location
        elif action == "go" and len(parts) > 1:
            self.move(parts[1])
        elif action in ["north", "south", "east", "west", "up", "down"]:
            self.move(action)
        elif action in ["fight", "attack"]:
            self.initiate_combat()
        elif action in ["get", "take", "pick"] and len(parts) > 1:
            self.pick_up_item(" ".join(parts[1:]))
        elif action == "use" and len(parts) > 1:
            self.player.use_item(" ".join(parts[1:]))
        elif action in ["inventory", "i"]:
            self.player.show_inventory()
        elif action == "status":
            print(f"\n  {self.player}")
        elif action == "quit":
            print("\n  Giving up? Goodbye.")
            self.game_running = False
        else:
            print("  Unknown command. Type 'help' for options.")

    def move(self, direction):
        if direction not in self.current_location.connections:
            print(f"  You can't go {direction} from here.")
            return
        next_loc = self.current_location.connections[direction]
        self.current_location = next_loc
        print(f"\n  You move {direction}...")
        self._show_room()                  # always reads live data
        if self.current_location == self.rooftop:
            self.state = Game.VICTORY
            return
        if self.current_location.enemies:
            print("\n  Enemies are here! Prepare to fight.")
            self.initiate_combat()

    def pick_up_item(self, item_name):
        if len(self.player.inventory) >= 5:
            print("  Your inventory is full (max 5 items). Use or drop something first.")
            return
        for item in self.current_location.items:
            if item_name.lower() in item.name.lower():
                self.player.pick_up(item)
                self.current_location.items.remove(item)
                return
        print(f"  There's no '{item_name}' here.")

    def initiate_combat(self):
        alive = [e for e in self.current_location.enemies if e.is_alive()]
        if not alive:
            print("  Nothing to fight here.")
            return
        enemy  = alive[0]
        battle = Combat(self.player, enemy)
        result = battle.start()

        if result == "victory":
            self.current_location.enemies.remove(enemy)
            self.player.gain_xp(enemy.xp_value)
            print(f"  XP gained: {enemy.xp_value}  |  Total XP: {self.player.xp}")

            # Drop loot into room (respect 5-item cap)
            for loot_item in enemy.loot:
                if len(self.current_location.items) < 5:
                    self.current_location.items.append(loot_item)
                    print(f"  {enemy.name} dropped: {loot_item.name}")

            # Unlock rooftop if Dean was just defeated
            if isinstance(enemy, ZombieDean):
                print("\n  The emergency roof door clicks open.")
                self.admin_building.add_connection("up", self.rooftop)

            # XP upgrade prompt
            self.xp_upgrade_prompt()

            # Show room state after combat so player sees new drops
            self._show_room()

        elif result == "defeat":
            self.state = Game.GAME_OVER

    # ------------------------------------------------------------------
    # XP Upgrade System
    # ------------------------------------------------------------------

    def xp_upgrade_prompt(self):
        """
        After a combat victory, let the player spend XP on permanent upgrades.
        Caps: +8 bonus STR total, +8 bonus max HP total.
        """
        p = self.player
        print(f"\n  ~~~ SPEND XP ~~~  (available: {p.xp} XP)")
        while True:
            str_left = Player.MAX_BONUS - p.str_bonus
            hp_left  = Player.MAX_BONUS - p.hp_bonus
            print(f"\n  [1] +1 STR          ({Player.STR_UPGRADE_COST} XP)  "
                  f"— current STR: {p.strength}  |  bonus left: {str_left}/{Player.MAX_BONUS}")
            print(f"  [2] +1 Max HP       ({Player.HP_UPGRADE_COST} XP)  "
                  f"— current max HP: {p.max_health}  |  bonus left: {hp_left}/{Player.MAX_BONUS}")
            print(f"  [3] Recover 10 HP   ({Player.HEAL_COST} XP)  "
                  f"— current HP: {p.health}/{p.max_health}")
            print(f"  [0] Done            (XP remaining: {p.xp})")
            choice = input("  > ").strip()

            if choice == "1":
                if p.xp < Player.STR_UPGRADE_COST:
                    print(f"  Not enough XP (need {Player.STR_UPGRADE_COST}).")
                elif str_left <= 0:
                    print(f"  STR is maxed out (cap: +{Player.MAX_BONUS}).")
                else:
                    p.xp        -= Player.STR_UPGRADE_COST
                    p.strength  += 1
                    p.str_bonus += 1
                    print(f"  STR increased to {p.strength}!  (XP remaining: {p.xp})")

            elif choice == "2":
                if p.xp < Player.HP_UPGRADE_COST:
                    print(f"  Not enough XP (need {Player.HP_UPGRADE_COST}).")
                elif hp_left <= 0:
                    print(f"  Max HP is maxed out (cap: +{Player.MAX_BONUS}).")
                else:
                    p.xp         -= Player.HP_UPGRADE_COST
                    p.max_health += 1
                    p.health      = min(p.health + 1, p.max_health)
                    p.hp_bonus   += 1
                    print(f"  Max HP increased to {p.max_health}!  (XP remaining: {p.xp})")

            elif choice == "3":
                if p.xp < Player.HEAL_COST:
                    print(f"  Not enough XP (need {Player.HEAL_COST}).")
                else:
                    heal   = min(10, p.max_health - p.health)
                    p.xp  -= Player.HEAL_COST
                    p.health += heal
                    print(f"  Recovered {heal} HP!  ({p.health}/{p.max_health})  (XP remaining: {p.xp})")

            elif choice == "0":
                break
            else:
                print("  Enter 1, 2, 3, or 0.")

    # ------------------------------------------------------------------
    # Help / End screens
    # ------------------------------------------------------------------

    def show_help(self):
        print("\n  COMMANDS")
        print("  --------")
        print("  go [dir] / [dir]  -- Move (north south east west up down)")
        print("  look              -- Show room, enemies, items, and exits")
        print("  fight             -- Attack an enemy here")
        print("  get [item]        -- Pick up an item (max 5 in inventory)")
        print("  use [item]        -- Use an item from inventory")
        print("  inventory / i     -- List carried items")
        print("  status            -- Show HP and stats")
        print("  help              -- Show this list")
        print("  quit              -- Exit the game")

    def show_game_over(self):
        print("\n" + "=" * 60)
        print("              F A I L E D   C O U R S E")
        print("=" * 60)
        if self.player:
            print(f"\n  {self.player.name} fell to the undead.")
            print(f"  XP earned: {self.player.xp}")
        print("\n  The zombies were too many. The campus is lost.")
        print("  Better luck next enrollment period...")
        print("=" * 60)

    def show_victory(self):
        print("\n" + "=" * 60)
        print("         D E G R E E   C O N F E R R E D")
        print("=" * 60)
        if self.player:
            print(f"\n  {self.player.name} bursts through the roof door.")
        print("\n  The helicopter drops from the clouds. You made it.")
        print("  Behind you, the campus fades into the smoke.")
        if self.player:
            print(f"\n  Final XP: {self.player.xp}")
        print("=" * 60)
        print("\n       Congratulations — you survived Dead Credits.\n")
````

## Reflection: OOP Design Decisions

> **Submission note**: Final version copied to `labs/LAB01_submission/reflection.md`.

---

The core of the game rests on a `Character` base class that captures everything shared by the player and every enemy: a name, hit points, strength, and defense. Putting those attributes — and the d20 `attack()` logic — in one place means I never have to re-write the same combat math for each creature type. Both `Player` and `Enemy` inherit `Character` and only add what makes them distinct. `Player` gets an inventory and experience points; `Enemy` gets an `xp_value` and a loot list. That is the inheritance principle in practice: share what is common, extend what is unique.

The clearest payoff came with `ZombieDean`. Every enemy uses `Character.attack()`, but the Dean needed critical-strike behavior — a roll of 18 or higher on the d20 pays out double dice. Rather than adding an `if isinstance(enemy, ZombieDean)` branch inside `Combat`, I overrode `attack()` directly on the `ZombieDean` class. The `Combat` loop calls `enemy.attack(player)` without knowing anything about what kind of enemy it is; Python's dynamic dispatch routes the call to the right method automatically. The same pattern applies to the flavor-text each zombie type prints when it attacks — override once, ignore everywhere else.

The trickiest design problem was the rooftop exit. The win condition requires the `Game` object to add a connection to `admin_building` after the `ZombieDean` is defeated inside `Combat`. That means `Game` must hold references to both `admin_building` and `rooftop` from the moment `create_world()` returns, so that `initiate_combat()` can reach in and call `admin_building.add_connection()`. Managing those cross-object references — deciding which object *owns* which piece of state — is one of the genuine design challenges in OOP. If I had more time I would refactor `create_world()` into a `World` class that `Game` holds as a single attribute, making those references cleaner and easier to extend with additional unlock events.
